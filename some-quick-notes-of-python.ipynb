- pandas is a Python library used for data manipulation and analysis.
- pd is the alias for Pandas.

# **Create DataFrame**

````python
data = {
    'feature1': [10, 20, 30],
    'feature2': [5, 15, 25]
}

df = pd.DataFrame(data)
````

- data → A dictionary containing the data.
- pd.DataFrame(data) → Creates a DataFrame from the dictionary.
- df → The variable that stores the DataFrame.

# **View Data**

````python
df.head()       # First 5 rows
df.tail()       # Last 5 rows
df.shape        # Number of rows and columns
df.columns      # Column names
df.index        # Row indexes
df.dtypes       # Data types of columns
df.info()       # DataFrame information
df.describe()   # Statistical summary
````

# **Select Rows & Columns**

````python
'''
df['feature1']                  # Select one column
df[['feature1', 'feature2']]    # Select multiple columns

df.iloc[0]       # Select the first row by position
df.iloc[0, 1]    # First row, second column
df.iloc[:, 0]    # All rows, first column

df.loc[0]                    # Select the row with index label 0
df.loc[0, 'feature1']        # Select a specific value
'''
````

## **Difference between loc and iloc:**
- loc → Selects data using index labels.
- iloc → Selects data using numerical positions.

# **Basic Statistical Operations**

````python
'''
df.mean()       # Mean of each numeric column
df.std()        # Standard deviation
df.median()     # Median
df.min()        # Minimum value
df.max()        # Maximum value
df.sum()        # Sum of values
df.count()      # Count of non-missing values
df.corr()       # Correlation between numeric columns

'''
````

## **To calculate statistics for a specific column:**

````python
'''
df['feature1'].mean()
df['feature1'].std()

'''
````

# **When a DataFrame is passed, Pandas performs these operations on each applicable column by default.**

# **Python Functions**

````python
def function_name(parameter):
    result = parameter
    return result
````

- def → Defines a function.
- function_name → Name of the function.
- parameter → Receives the input value.
- return → Sends the result back to the caller.

# **Example: Standardization**

````python
'''
def standardize_data(data):
    mean = data.mean()
    std_dev = data.std()
    result = (data - mean) / std_dev
    return result

X_scaled = standardize_data(df)

'''
````

- Passing df → Processes all applicable columns.
- Passing df['feature1'] → Processes only feature1.
- mean() → Calculates the mean.
- std() → Calculates the sample standard deviation by default.

# **Filter Data**

````python
'''
df[df['feature1'] > 10]       # Values greater than 10
df[df['feature1'] == 10]      # Values equal to 10
df[df['feature1'] != 10]      # Values not equal to 10
df[df['feature1'] <= 10]      # Values less than or equal to 10

'''
````

# **Multiple conditions**

````python
'''
df[(df['feature1'] > 10) & (df['feature2'] < 20)]
df[(df['feature1'] > 10) | (df['feature2'] < 20)]

'''
````

- & → AND
- | → OR
- == → Equal to
- != → Not equal to
- ``` > → Greater than ```
- < → Less than

##### Important: Use parentheses around each condition when combining Pandas conditions with & or |.

# **Important Concepts to Remember**


## 14. Important Concepts

### A. Variable

```python
X = df
```

`X` refers to the same DataFrame object as `df`.

### B. Parameter and Argument

```python
def calculate_mean(data):   # data is a parameter
    return data.mean()

calculate_mean(df)          # df is the argument
```

- **Parameter:** The variable defined in the function.
- **Argument:** The value passed to the function when calling it.

### C. DataFrame vs Series

```python
df['feature1']       # Series: one column
df[['feature1']]     # DataFrame: one-column DataFrame
df                   # DataFrame: multiple columns
```

- **Series:** A one-dimensional data structure.
- **DataFrame:** A two-dimensional data structure containing rows and columns.

### D. Column-Wise Operations

```python
df.mean()
```

Calculates the mean of each applicable column separately.

### E. Standardization Formula

$$
z = \frac{x-\mu}{\sigma}
$$

- `x` → Original value
- `μ` → Mean of the feature
- `σ` → Standard deviation of the feature
- `z` → Standardized value

**Example:**

```python
X = df[['feature1', 'feature2']]

X_scaled = (X - X.mean()) / X.std()
```

Both features are standardized independently because `X` contains both columns.

---

## Final Rule to Remember

**The object passed to a function determines which data it processes.**

- **DataFrame** → Multiple columns.
- **Series** → A single column.
- **Pandas operations** generally work column-wise on a DataFrame by default.
